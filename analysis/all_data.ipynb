```python
from pandas import DataFrame
import seaborn as sns
from data_analysis import read_all_data
import matplotlib.pyplot as plt
```

```python
df: DataFrame = read_all_data()
print(df.columns)
print(df.describe())
print(df["adversary"].value_counts())
print(df["player"].value_counts())
```

```python
ci = 99
without_random_df = df[~(df["adversary"] == "random")]
g1 = sns.catplot(
    data=without_random_df,
    x="player",
    y="score",
    errorbar=("ci", ci),
    hue="adversary",
    palette="bright",
    kind="bar",
)
g1.figure.suptitle(
    f"Average Scores Between Algorithms with {ci}% Confidence Interval", y=1.03
)

df: DataFrame = read_all_data()
with_random_df = df[(df["adversary"] == "random")]
g1 = sns.catplot(
    data=with_random_df,
    x="player",
    y="score",
    errorbar=("ci", ci),
    hue="adversary",
    palette="bright",
    kind="bar",
)
g1.figure.suptitle(
    f"Average scores for Algorithms Against Random Adversary\nwith {ci}% Confidence Interval", y=1.03
)
```
